```python
str = input("Enter your state of origin:")

print ("Your state of origin is: ", str)
print ("The first character is: ", str[0])
print ("The characters starting from 3rd to 5th are: ", str[2:5])
print ("The string starting from 3rd character is: ", str[2:])
print ("State of origin two times", str * 2)
```

```python
# Input from user
m = float(input("Enter mass in kilogram: "))

# Constant value for the speed of light in m/s
c = 299792458

# Calculating energy using Einstien's equation
energy = m * c ** 2

# Displaying the result
print (f"The energy equivalent to {m} kg of mass is {energy} joules.")
```

```python
my_list = [ 'Anaconda', 786, 2.23, 'Jupyter', 70.2 ]
shortlist = [321, 'Python', 'Anaconda']

print(my_list)        # Prints complete line
print(my_list[0])     # print first element of the list
print(my_list[1:3])   # print elements starting from 2nd till 3rd
print(my_list[2:])    # Print elements starting from 3rd element
print(shortlist * 2)   # print list 2 times
print(my_list + shortlist) # print concatenatted lists
```

```python
# Returns false as game1 is not equal to game2
game1 = 2
game2 = 4
print(bool(game1 == game2))
# Or
print(game1 == game2)

# Returns False as val is None
val = None
print(bool(val))

# Returns false as num is an empty sequence
num = ()
print(bool(num))

# Returns true as age as boolean
age = True
print(bool(age))
```

```python
tuple = ('Ekiti', 750, 'Oshogbo', 250, 'Akure', 500)
s_tuple = ('Abeokuta', 300, 'Ogbomoso')

# Prints last complete tuple
print(tuple)

# Prints the last element of the tuple
print(tuple[-1])

# Prints elements of the tuple starting from the 2nd the 3rd
print(tuple[2:4])

# Prints the elements of the tuple starting from the 3rd element
print(tuple[3:])

# Prints the contents of the entire tuple twice
print(s_tuple * 2)

# Prints concatenated tuples
print(tuple + s_tuple)
```

```python
# Convert to int

grade = int(70)   # Grade will be 70
gpa = int(4.9)    # gpa will be 4
cgpa = int("4")   # cgpa will be 4

print(f"Grade = {grade}")
print(f"GPA = {gpa}")
print(f"CGPA = {cgpa}")
```

```python
# Converts to floats


grade = float(97)
gpa = float(5)
cgpa = float("4.7")

print(f"grade = {grade}")
print(f"GPA = {gpa}")
print(f"CGPA = {cgpa}")
```

```python
# Print the given sequence of numbers using for loop
count = [10, 20, 30, 40, 50]
for num in count:
    print(num)
```

```python
# Prints multiples of 10 for nummbers in a given range
for num in range(5):
    if num > 0:
        print(num * 10)
```

```python
for num in range(1,5): # Starts from 1, excludes 5.
    print(num * 10)
```

```python
# Print natural numbers from 5 and up using while loop, if and break statement
count = 5
while count >= 5:
    print(count)
    count += 1
    if count == 20:
        break
```

```python
# Print first 5 natural numbers using while loop
count = 1
while count <= 5:
    print(count)
    count += 1
```
