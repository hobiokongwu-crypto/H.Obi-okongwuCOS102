```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('AB_NYC_2019.csv')

df.columns = df.columns.str.strip().str.lower()

cols_to_drop =[' id','	name','	host_id','	host_name','	neighbourhood_group','	neighbourhood','	latitude','	longitude','	room_type','	price']
df.drop(columns=cols_to_drop, inplace=True, errors='ignore')

df['name'] = df['name'].fillna('Unknown')

df = df[(df['price'] >= 10) & (df['price'] <= 1000)]

print("Columns currently in your data:", df.columns.tolist())
```

```python
plt.figure(figsize = (10, 6))
sns.barplot(x = 'neighbourhood_group', y = 'price', data = df, estimator = 'mean', errorbar = None)

plt.title('Average Nightly Price by Borough')
plt.xlabel('Borough (Neighborhood Group)')
plt.ylabel('Average Price ($)')
plt.show()
```

Insight: Manhattan has the highest average price, with Brooklyn being a close second. This confirms that Manhattan remains the premium tourism hub of NYC, while the Bronx offers the most budget-friendly stays.

```python
room_counts = df['room_type'].value_counts()
plt.figure(figsize = (8, 8))
plt.pie(room_counts, labels = room_counts.index, autopct = '%1.1f%%', colors = ['#ff9999', '#66b3ff', '#99ff99'])
plt.title('Distribution of Airbnb Room Types in NYC')
plt.show()
```

Insight: Entire homes/apartments and Private rooms make up over 50% of the market.This suggests that travelers in NYC prioritize privacy over "shared" living spaces, which account for a tiny fraction of listings.

```python
plt.figure(figsize = (10, 6))
sns.scatterplot(x = 'availability_365', y = 'price', hue = 'neighbourhood_group', data = df[df['price'] < 500], alpha = 0.4)
plt.title('Price vs. Yearly Availability (Prices < $500)')
plt.show()
```

Insight: There is a heavy concentration of listings with very low availability (near 0) across all price points. This indicates a high number of "part-time" hosts who likely only rent out their homes while they are away, rather than running full-time businesses.
