```python
import pandas as pd
df = pd.read_csv('SIS.csv')
df

class StudentClassifier:
    def __init__(self, filepath):
        self.__filepath = filepath
        self.__data = None
        self.__pirates = None
        self.__yankees = None
        self.__bulls = None
        self.__over25_error = None

    def get_pirates(self):
        return self.__pirates

    def get_yankees(self):
        return self.__yankees

    def get_bulls(self):
        return self.__bulls

    def get_over25_error(self):
        return self.__over25_error

    def load_data(self):
        self.__data = pd.read_csv(self.__filepath)
        print("SIS data loaded successfully.\n")
        print("=== Full Student Record ===")
        print(self.__data.to_string(index=False))
        print()

    def classify(self):
        self.__pirates = self.__data[
            (self.__data['Age'] > 14) & (self.__data['Age'] < 18)
        ].copy()

        self.__yankees = self.__data[
            (self.__data['Age'] > 18) & (self.__data['Age'] < 22)
        ].copy()

        self.__bulls = self.__data[
            (self.__data['Age'] > 22) & (self.__data['Age'] < 25)
        ].copy()

        over25 = self.__data[self.__data['Age'] > 25]
        if not over25.empty:
            names = ', '.join(over25['Name'].tolist())
            self.__over25_error = (
                f"ERROR: The following student(s) do not fall within any "
                f"valid age category (age > 25): {names}. "
                f"Request encapsulated — access via get_over25_error()."
            )

    def save_and_display(self):
        categories = {
            'The_Pirates': self.__pirates,
            'The_Yankees': self.__yankees,
            'The_Bulls':   self.__bulls,
        }

        for name, df in categories.items():
            print(f"=== {name.replace('_', ' ')} ===")
            if df.empty:
                print("  No students in this category.\n")
            else:
                print(df.to_string(index=False))
                df.to_csv(f"{name}.csv", index=False)
                print(f"  Saved to '{name}.csv'\n")

        error_msg = self.get_over25_error()
        if error_msg:
            print("=== Students Aged > 25 (Encapsulated) ===")
            print(f"  {error_msg}\n")

    def run(self):
        self.load_data()
        self.classify()
        self.save_and_display()


# Run it
classifier = StudentClassifier("SIS.csv")
classifier.run()
```
