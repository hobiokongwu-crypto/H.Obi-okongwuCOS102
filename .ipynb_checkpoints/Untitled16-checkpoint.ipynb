```python
class SST:
    prog1 = 'Computer Science'

    def __init__(self, name) :
        self.name = name
    def speak(self) :
        print('\nMy Name is', self.name)
        print('I study', SST.prog1)


stud1 = SST(input('Enter student 1 name : '))
stud2 = SST(input('Enter student 2 name : '))

stud1.speak()
stud2.speak()
```
