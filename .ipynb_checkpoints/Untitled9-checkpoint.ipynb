```python
# ATR calculator
s_xp = int(input("How many years of experience do you have?:")) # Years of experience of staff member
s_age = int(input("How old are you?:"))# Age of staff members

# Conditions for age and experience
if s_xp > 25 and s_age >= 55:
    atr = 5600000
    print(f"Your Annual Tax Revenue is: N{atr}")
elif s_xp > 20 and s_age >= 45:
    atr = 4480000
    print(f"Your Annual Tax Revenue is: N{atr}")
elif  s_xp > 10 and s_age >= 35:
    atr = 1500000
    print(f"Your Annual Tax Revenue is: N{atr}")
else :
    atr = 550000
    print(f"Your Annual Tax Revenue is: N{atr}")
```
