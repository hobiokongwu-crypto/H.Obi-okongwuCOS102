```python
# Formula construction
def calc_simple_interest(principal, rate, time):
    simple_interest = (principal * rate * time)/ 100
    return simple_interest

# Parameter input focus
P = float(input("Enter principal amount: "))
R = float(input("Enter annual interest rate (%): "))
T = float(input("Enter the time in years: "))

# The closing end
interest = calc_simple_interest(P, R, T)
print(f"Your Simple Interest is: {interest}")
```

```python
# Formula construction
def calc_compound_interest(principal, rate, time, compound_interests_per_year):
    rate_decimal = rate / 100.0
    amount = principal * ( 1.0 + (rate_decimal / compound_interests_per_year)) ** (compound_interests_per_year * time)
    compound_interest = amount - principal
    return compound_interest

# Parameter input focus
P = float(input("Enter the principal amount: "))
R = float(input("Enter the annual interest rate: "))
t = float(input("Enter the number of years that went by: "))
n = float(input("Enter the number of times the interest was compounded per year: "))

# Final conclusion to the calculation
interest = calc_compound_interest(P, R, t, n)
print(f"Your Compound Interest is: {interest}")
```

```python
# Formula construction
def calc_annuity_plan(equal_payment_series, rate, compound_interests_per_year, time):
    rate_decimal = rate / 100.0
    segment_a = (((1 + (rate_decimal / compound_interests_per_year)) ** (compound_interests_per_year * time)) - 1)
    annuity_plan = (equal_payment_series * segment_a) / (rate_decimal / compound_interests_per_year)
    return annuity_plan

# Formula parameter input
PMT = float(input("Enter the series of equal payments: "))
R = float(input("Enter the annual interest rate: "))
n = float(input("Enter the number of times the interest was compounded per year: "))
t = float(input("Enter the number of years that went by: "))

# The end result of the calculation
A = calc_annuity_plan(PMT, R, n, t)
print(f"The future value of the annuity is: {A}")
```
