```python
import tkinter as tk
from tkinter import messagebox
import random

# Employees and Tsk Lists
employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu", "Stella Mankinde", "Jane Akibo", "Ago James",
    "Michell Taiwo", "Abraham Jones", "Nicole Anide", "Kosi Korso", "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"]

tasks = ["Loading", "Transporting", "Reviewing Orders", "Customer Service", "Delivering Items"]

# OOP CLass
class Employee:
    def __init__ (self, name):
        self.name = name

    def check_employee(self):
        return self.name in employees

    def take_attendance(self):
        return f"Attendance  taken for {self.name}"

    def refuse_access(self):
        return "Access Denied! You are not an employee."

    def assign_task(self):
        task = random.choice(tasks)
        return f"Task assigned: {task}"
# GUI Function
def process():
    name = entry.get()
    emp = Employee(name)

    if emp.check_employee():
        attendance = emp.take_attendance()
        task = emp.assign_task()
        messagebox.showinfo("Success", f"{attendance}\n{task}")
    else:
        messagebox.showerror("Error", emp.refuse_access())

# GUI Setup
root = tk.Tk()
root.title("Employees Attendance System")
root.geometry("400x200")

label = tk.Label(root, text = "Enter Employee Name:")
label.pack(pady = 10)

entry = tk.Entry(root, width = 30)
entry.pack(pady = 5)

button = tk.Button(root, text = "Submit", command = process)
button.pack(pady = 20)


root.mainloop()
    
```
