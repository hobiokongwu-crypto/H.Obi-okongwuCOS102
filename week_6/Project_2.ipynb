```python
import tkinter as tk
from tkinter import messagebox

#OOP Class
class Delivery:
    def __init__(self, location , weight):
        self.location = location.lower()
        self.weight = weight

    def calculate_fee(self):
        if self.location == "pau":
            if self.weight >= 10:
                return 2000
            else:
                return 1500

        elif self.location == "epe":
            if self.weight >= 10:
                return 5000
            else:
                return 4000

        elif self.location == "lekki":
            if self.weight >= 10:
                return 8000
            else:
                return 6500

        else:
            return None

    def invalid_location(self):
        return "Invalid Location! Only PAU, Epe, or Lekki allowed."

# GUI Function
def calculate():
    location = entry_location.get()

    try:
        weight = float(entry_weight.get())
    except ValueError:
        messagebox.showerror("Error", "Please enter a valid number for weight.")
        return

    delivery = Delivery(location, weight)
    fee = delivery.calculate_fee()

    if fee is not None:
        messagebox.showinfo("Delivery Fee", f"Amount to pay: ₦{fee}")
    else:
        messagebox.showerror("Error", delivery.invalid_location())

# GUI Setup
root = tk.Tk()
root.title("Delivery Fee Calculator")
root.geometry("400x250")

#Location
label1 = tk.Label(root, text = "Enter Location (PAU or Epe or Lekki):")
label1.pack(pady = 10)

entry_location = tk.Entry(root, width = 30)
entry_location.pack(pady = 5)

#Weight
label2 = tk.Label(root, text = "Enter Package Weight (kg):")
label2.pack(pady = 10)

entry_weight = tk.Entry(root, width = 30)
entry_weight.pack(pady = 5)

#Button
button = tk.Button(root, text = "Calculate Fee", command = calculate)
button.pack(pady = 20)

root.mainloop()
```
